```python
# ==============================================================================
# 🔍 설치된 38개 AI & Data Science 패키지 전수 임포트 검증
# ==============================================================================
import importlib

# (패키지명, 실제 import할 모듈명)
PACKAGES = [
    ("beautifulsoup4", "bs4"),
    ("colorama", "colorama"),
    ("deepagents", "deepagents"),
    ("docling", "docling"),
    ("fastapi", "fastapi"),
    ("flask", "flask"),
    ("html5lib", "html5lib"),
    ("ipykernel", "ipykernel"),
    ("jupyter", "jupyter"),
    ("kiwipiepy", "kiwipiepy"),
    ("langchain", "langchain"),
    ("langchain-chroma", "langchain_chroma"),
    ("langchain-classic", "langchain_classic"),
    ("langchain-community", "langchain_community"),
    ("langchain-core", "langchain_core"),
    ("langchain-daytona", "langchain_daytona"),
    ("langchain-docling", "langchain_docling"),
    ("langchain-openai", "langchain_openai"),
    ("langchain-tavily", "langchain_tavily"),
    ("langchain-text-splitters", "langchain_text_splitters"),
    ("langgraph", "langgraph"),
    ("langgraph-checkpoint-postgres", "langgraph_checkpoint_postgres"),
    ("langsmith", "langsmith"),
    ("llama-cloud", "llama_cloud"),
    ("lxml", "lxml"),
    ("markdownify", "markdownify"),
    ("matplotlib", "matplotlib"),
    ("numpy", "numpy"),
    ("openai", "openai"),
    ("pandas", "pandas"),
    ("psycopg", "psycopg"),
    ("pypdf", "pypdf"),
    ("python-dotenv", "dotenv"),
    ("python-telegram-bot", "telegram"),
    ("rank-bm25", "rank_bm25"),
    ("requests", "requests"),
    ("scikit-learn", "sklearn"),
    ("sentence-transformers", "sentence_transformers"),
    ("slack-sdk", "slack_sdk"),
    ("tabulate", "tabulate"),
    ("tavily", "tavily"),
    ("torch", "torch"),
]

print("=" * 65)
print(f"{'패키지명':<32} | {'상태':<6} | {'버전'}")
print("=" * 65)

success_count = 0
for pkg_name, module_name in PACKAGES:
    try:
        mod = importlib.import_module(module_name)
        ver = getattr(mod, "__version__", "정상 로드됨")
        print(f"{pkg_name:<32} | ✅ OK   | {ver}")
        success_count += 1
    except Exception as e:
        print(f"{pkg_name:<32} | ❌ FAIL | {e}")

print("=" * 65)
print(f"🎉 총 {success_count} / {len(PACKAGES)} 개 패키지가 정상 작동합니다!")
print("=" * 65)
```
